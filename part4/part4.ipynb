```python
import subprocess
import sys

def predict_with_sampling(top_k, top_p, temperature):
    cmd = [
        sys.executable, "predict.py",
        "--sentence", "Two boys are sitting on a couch that is red and black.",
        "--checkpoint-path", "../part4/nmt.model",
        "--model-config", "../part4/model_config.yaml",
        "--vocab-file", "vocab/vocab.json",
        "--do-sample",
        "--top-k", str(top_k),
        "--top-p", str(top_p),
        "--temperature", str(temperature),
    ]

    proc = subprocess.Popen(
        cmd,
        cwd="../part2/",
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1,
        universal_newlines=True
    )

    # Read output line-by-line and print it
    for line in proc.stdout:
        if "Prediction:" in line:
            print(line.strip())
```

```python
predict_with_sampling(1, 1, 1)
```

```python
top_ks = [5, 20, 50]
top_ps = [0.5, 0.9, 0.95]
temperatures = [0.7, 1.0, 1.3]

```

```python
for top_k in top_ks:
    for top_p in top_ps:
        for temperature in temperatures:
            print(f"\n--- top_k={top_k}, top_p={top_p}, temperature={temperature} ---")
            predict_with_sampling(top_k, top_p, temperature)

```

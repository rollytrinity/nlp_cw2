```python
english = []
french = []

with open('multi30k_data/train.en', 'r', encoding='utf-8') as en_file, \
     open('multi30k_data/train.fr', 'r', encoding='utf-8') as de_file:
    for en_line, fr_line in zip(en_file, de_file):
        en_line = en_line.strip()
        de_line = fr_line.strip()
        
        english.append(en_line)
        french.append(de_line)
```

```python
def count_word(word, sentences):
    return len([s for s in sentences if word in s])
```

```python
colours = ['red', 'blue', 'green', 'yellow', 'black', 'white', 'orange', 'purple', 'pink', 'brown']
french_colours = [
    "rouge",     # red
    "orange",    # orange
    "jaune",     # yellow
    "vert",      # green
    "bleu",      # blue
    "violet",    # purple
    "rose",      # pink
    "marron",    # brown
    "noir",      # black
    "blanc",     # white
    "gris",      # grey
    "beige",     # beige
    "turquoise",
    "kaki",
    "indigo",
    "pourpre"
]
```

```python
vetus_count = sum([count_word('vêtus de ' + colour, french) for colour in french_colours])
est_count = sum([count_word('est ' + colour, french) for colour in french_colours])
```
